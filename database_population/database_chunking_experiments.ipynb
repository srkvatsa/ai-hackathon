```python
import os
from langchain.text_splitter import RecursiveCharacterTextSplitter

def chunk_document(text, chunk_size, chunk_overlap):
    splitter = RecursiveCharacterTextSplitter(chunk_size=chunk_size, chunk_overlap=chunk_overlap)
    return splitter.split_text(text)

def write_chunks_to_files(text, output_dir="chunks", chunk_size=256, chunk_overlap=50):
    """Splits text into chunks and writes each to a separate file."""
    os.makedirs(output_dir, exist_ok=True)  # Ensure the directory exists

    chunks = chunk_document(text, chunk_size, chunk_overlap)  # Split text

    for i, chunk in enumerate(chunks):
        file_path = os.path.join(output_dir, f"chunk_{i}.txt")
        with open(file_path, "w", encoding="utf-8") as f:
            f.write(chunk)

    print(f"Saved {len(chunks)} chunks to {output_dir}")

def load_chunks_from_files(input_dir="chunks"):
    """Loads text chunks from saved files."""
    chunks = []
    for filename in sorted(os.listdir(input_dir)):  # Ensure order
        if filename.startswith("chunk_") and filename.endswith(".txt"):
            file_path = os.path.join(input_dir, filename)
            with open(file_path, "r", encoding="utf-8") as f:
                chunks.append(f.read())
    return chunks

# Example usage

# Modify chunk size and overlap as needed
chunk_size = 512
chunk_overlap = 100

f = open('data/roman-history.txt', 'r')
res = f.read()
print(res)
write_chunks_to_files(res, chunk_size=chunk_size, chunk_overlap=chunk_overlap)
loaded_chunks = load_chunks_from_files()
print(f"Loaded {len(loaded_chunks)} chunks.")
```
